```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("ggplot")
```

```python
performance = pd.read_csv("../data/processed/clean_performance.csv")

performance.head()
```

```python
print("Rows :", performance.shape[0])
print("Columns :", performance.shape[1])

performance.info()
```

```python
performance.describe()
```

```python
top_1yr = (
    performance
    .sort_values("return_1yr_pct", ascending=False)
    .head(10)
)

top_1yr[["scheme_name", "return_1yr_pct"]]
```

```python
top_3yr = (
    performance
    .sort_values("return_3yr_pct", ascending=False)
    .head(10)
)

top_3yr[["scheme_name", "return_3yr_pct"]]
```

```python
top_5yr = (
    performance
    .sort_values("return_5yr_pct", ascending=False)
    .head(10)
)

top_5yr[["scheme_name", "return_5yr_pct"]]
```

```python
top_sharpe = (
    performance
    .sort_values("sharpe_ratio", ascending=False)
)

top_sharpe[
    ["scheme_name", "sharpe_ratio"]
].head(10)
```

```python
plt.figure(figsize=(8,5))

plt.hist(performance["expense_ratio_pct"], bins=20)

plt.title("Expense Ratio Distribution")
plt.xlabel("Expense Ratio (%)")
plt.ylabel("Frequency")

plt.show()
```

```python
expense = (
    performance
    .groupby("category")["expense_ratio_pct"]
    .mean()
    .sort_values(ascending=False)
)

expense
```

```python
expense.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average Expense Ratio by Category")

plt.ylabel("Expense Ratio (%)")

plt.tight_layout()

plt.show()
```

```python
performance["risk_grade"].value_counts()
```

```python
performance["risk_grade"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Risk Grade Distribution")
plt.ylabel("Number of Funds")

plt.show()
```

```python
performance["morningstar_rating"].value_counts().sort_index()
```

```python
performance["morningstar_rating"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Morningstar Ratings")
plt.xlabel("Rating")
plt.ylabel("Funds")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    performance["std_dev_ann_pct"],
    performance["return_3yr_pct"]
)

plt.xlabel("Standard Deviation (%)")
plt.ylabel("3-Year Return (%)")

plt.title("Risk vs Return")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    performance["beta"],
    performance["alpha"]
)

plt.xlabel("Beta")
plt.ylabel("Alpha")

plt.title("Alpha vs Beta")

plt.show()
```

```python
category_return = (
    performance
    .groupby("category")["return_3yr_pct"]
    .mean()
    .sort_values(ascending=False)
)

category_return
```

```python
category_return.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average 3-Year Return by Category")

plt.ylabel("Return (%)")

plt.tight_layout()

plt.show()
```

## Key Insights

- Some schemes consistently outperform others across 1-year, 3-year, and 5-year periods.
- Higher Sharpe ratios indicate better risk-adjusted performance.
- Expense ratios vary across fund categories.
- Most schemes are concentrated in Moderate and High risk grades.
- Higher risk does not always result in higher returns.
- Alpha and Beta provide additional insights into fund performance relative to the benchmark.
